## Repo And Scenario Orientation

The canonical scenario is a support triage and response drafting agent for operational tickets. The zero-key runtime path lets the notebook stay shareable without binding the lesson to live provider credentials.

```python
from llm_ops_v1.agents.base_agent import SupportTriageDependencies, run_support_triage_agent

prompt = (
    "A shipment is delayed. Draft the support reply, mention the tracking update, "
    "and decide whether to escalate."
)
deps = SupportTriageDependencies(
    ticket_id="ticket-demo-042",
    customer_tier="priority",
    ticket_priority="high",
)
triage_output = await run_support_triage_agent(prompt, deps=deps)
triage_output
```

## Cost And Routing

This cell keeps the lesson repo-first by using the pricing and local-infra helpers already in the package. It is enough to talk through hosted cost, cached tokens, and the difference between request pricing and self-hosted runtime cost.

```python
from dataclasses import asdict

from llm_ops_v1.economics.commercial_models import get_pricing
from llm_ops_v1.economics.cost_calculator import estimate_token_cost
from llm_ops_v1.economics.local_models import estimate_local_infra_cost

hosted_breakdown = estimate_token_cost(
    get_pricing("openai:gpt-4.1-mini"),
    input_tokens=1_200,
    output_tokens=350,
    cached_input_tokens=800,
)
local_infra_estimate = estimate_local_infra_cost("ollama:qwen2.5", latency_seconds=1.4)

{
    "hosted_breakdown": asdict(hosted_breakdown),
    "local_infra_estimate_usd": local_infra_estimate,
}
```

## Context, Memory And Caching

The notebook should show memory as constrained state, not magic. Here we use the in-memory session and episodic primitives plus the existing dashboard demo estimate to make caching visible without inventing a second implementation.

```python
from llm_ops_v1.dashboard.demo_runner import estimate_demo_metrics
from llm_ops_v1.memory.episodic import EpisodicMemory
from llm_ops_v1.memory.short_term import InMemorySessionState

session_state = InMemorySessionState()
session_state.put(deps.ticket_id, "customer_tier", deps.customer_tier)
session_state.put(deps.ticket_id, "ticket_priority", deps.ticket_priority)

episodes = EpisodicMemory()
episodes.record(
    actor="support-agent",
    action="drafted_reply",
    summary="The agent produced a zero-key reply draft for the delayed shipment.",
)

cache_estimate = estimate_demo_metrics(prompt=prompt, output=triage_output)

{
    "session_snapshot": session_state.snapshot(deps.ticket_id),
    "recent_episodes": [episode.summary for episode in episodes.latest()],
    "cache_hit_ratio": cache_estimate.cache_hit_ratio,
    "estimated_cost_usd": cache_estimate.total_cost_usd,
}
```

## Evals And LLM-as-a-Judge

For the notebook, the zero-key path is enough to explain what a rubric-driven score looks like before introducing a real judge model.

```python
from llm_ops_v1.dashboard.demo_runner import judge_zero_key_output

judge_score = judge_zero_key_output(prompt=prompt, output=triage_output)
judge_score.model_dump()
```

## Dashboard Interpretation

The dashboard is the control surface, not the source of truth. This cell reuses the seeded fixture rows and the existing live-record builder so the notebook can explain where the dashboard signals come from.

```python
from llm_ops_v1.dashboard.demo_runner import build_fixture_records, build_live_record
from llm_ops_v1.dashboard.store import DashboardStore

store = DashboardStore(build_fixture_records())
store.append(build_live_record(prompt=prompt, output=triage_output, latency_ms=900))

[record.table_row() for record in store.snapshot()[:3]]
```
